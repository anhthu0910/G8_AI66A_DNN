```python
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks, metrics
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_fscore_support
from sklearn.utils.class_weight import compute_class_weight
import os
import hashlib
from collections import Counter
```

```python
DATA_DIR = Path("/home/pc/AAA/project/deepnn/G8_AI66A_DNN/data/raw") # Data đã loại ảnh trùng từ file model 1
TRAIN_DIR = DATA_DIR / "train"
VAL_DIR = DATA_DIR / "val"
TEST_DIR = DATA_DIR / "test"

SETS = ["train", "val", "test"]
IMG_SIZE = 128
BATCH_SIZE = 32
EPOCHS = 40
BASE_CH = 32
DROPOUT = 0.3
MAX_LR = 2e-3
WEIGHT_DECAY = 5e-4
NUM_CLASSES = 4
```

```python
# Bật Mixed Precision để tăng tốc độ huấn luyện trên GPU
tf.keras.mixed_precision.set_global_policy('mixed_float16')

print(f"TensorFlow version: {tf.__version__}")
print(f"GPU Available: {tf.config.list_physical_devices('GPU')}")
```

## 1. Data preprocessing

Không cần loại ảnh trùng nữa vì đã làm từ file model 1 và đã lưu vào folder ban đầu

Lấy ảnh từ folder raw là những data đã được loại bỏ dup

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR, 
    image_size=(IMG_SIZE, IMG_SIZE), 
    batch_size=BATCH_SIZE,
    label_mode="categorical", 
    shuffle=True, 
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR, 
    image_size=(IMG_SIZE, IMG_SIZE), 
    batch_size=BATCH_SIZE,
    label_mode="categorical", 
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR, 
    image_size=(IMG_SIZE, IMG_SIZE), 
    batch_size=BATCH_SIZE,
    label_mode="categorical", 
    shuffle=False
)

class_names = ['CNV', 'DME', 'DRUSEN', 'NORMAL']
NUM_CLASSES = 4
```

```python
normalization_layer = tf.keras.layers.Normalization(axis=-1, dtype='float32')
# Chỉ truyền ảnh vào adapt(), không truyền nhãn
train_images_for_norm = train_ds.map(
    lambda images, labels: tf.cast(images, tf.float32)
)
normalization_layer.adapt(train_images_for_norm) # Tự động tính mean và variance
```

```python
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)
```

```python
# Định nghĩa ResBlock
def ResBlock(x, out_ch, stride=1):
    in_ch = x.shape[-1]
    shortcut = x

    # Shortcut 
    if stride != 1 or in_ch != out_ch:
        shortcut = layers.Conv2D(out_ch, 1, strides=stride, use_bias=False)(shortcut)
        shortcut = layers.BatchNormalization()(shortcut)

    # Conv 1
    y = layers.Conv2D(out_ch, 3, strides=stride, padding='same', use_bias=False)(x)
    y = layers.BatchNormalization()(y)
    y = layers.ReLU()(y)

    # Conv 2
    y = layers.Conv2D(out_ch, 3, strides=1, padding='same', use_bias=False)(y)
    y = layers.BatchNormalization()(y)

    
    y = layers.Add()([y, shortcut])
    y = layers.ReLU()(y)
    return y


# Xây dựng ResNet
def build_resnet(input_shape, num_classes, base_ch=32, dropout=0.3):
    inputs = layers.Input(shape=input_shape)

    x = layers.RandomFlip("horizontal")(inputs)
    x = layers.RandomRotation(15 / 360)(x)
    x = layers.RandomZoom((-0.1, 0.1))(x)
    x = layers.RandomBrightness(0.2)(x)
    x = layers.RandomContrast(0.2)(x)
    
    # Normalization
    x = normalization_layer(x)
    
    # Stem
    x = layers.Conv2D(base_ch, 3, strides=1, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    # Stage 1 (base_ch * 1)
    x = ResBlock(x, base_ch * 1, stride=1)
    x = ResBlock(x, base_ch * 1, stride=1)

    # Stage 2 (base_ch * 2)
    x = ResBlock(x, base_ch * 2, stride=2)
    x = ResBlock(x, base_ch * 2, stride=1)

    # Stage 3 (base_ch * 4)
    x = ResBlock(x, base_ch * 4, stride=2)
    x = ResBlock(x, base_ch * 4, stride=1)

    # Stage 4 (base_ch * 8)
    x = ResBlock(x, base_ch * 8, stride=2)
    x = ResBlock(x, base_ch * 8, stride=1)

    # Head
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(dropout)(x)
    
    # Output (Cast to float32 để tương thích mixed_precision)
    outputs = layers.Dense(num_classes, activation='softmax', dtype='float32')(x)

    return tf.keras.Model(inputs, outputs)

model = build_resnet((IMG_SIZE, IMG_SIZE, 3), NUM_CLASSES, BASE_CH, DROPOUT)
```

```python
model.summary()
```

```python
model.compile(
    optimizer=optimizers.Adam(learning_rate=MAX_LR, weight_decay=WEIGHT_DECAY),
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
    metrics=['accuracy', metrics.F1Score(average='macro', name='f1_score')]
)

# Callbacks
cb_list = [
    callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
    callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6),
]
```

```python
class_counts = [31736, 11166, 8213, 50328] 
class_weights = compute_class_weight(
    class_weight='balanced', 
    classes=np.arange(NUM_CLASSES), 
    y=np.repeat(np.arange(NUM_CLASSES), class_counts)
)
class_weight_dict = dict(enumerate(class_weights))
print("Class weights:", class_weight_dict)
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=cb_list,
    class_weight=class_weight_dict
)
```

## 5. Danh gia tren tap test

```python
y_true = np.concatenate([y for x, y in test_ds], axis=0)
y_true = np.argmax(y_true, axis=1)

y_pred_probs = model.predict(test_ds)
y_pred = np.argmax(y_pred_probs, axis=1)
```

```python
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# Vẽ Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
